# Credit Card Fraud Detection: EDA
Explore the transaction dataset, with attention to its highly imbalanced fraud target.

Expected input: `../data/creditcard.csv`, with `Class` equal to 1 for fraud.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_CANDIDATES = (Path('data/creditcard.csv'), Path('../data/creditcard.csv'), Path('credit-card-fraud-detection/data/creditcard.csv'))
DATA_PATH = next((path for path in DATA_CANDIDATES if path.exists()), DATA_CANDIDATES[0])
data = pd.read_csv(DATA_PATH)
data.head()
```

```python
print(f'Shape: {data.shape}')
display(data.dtypes.to_frame('dtype'))
display(data.isna().sum().sort_values(ascending=False).head(10).to_frame('missing'))
display(data.describe().T)
```

```python
class_counts = data['Class'].value_counts().sort_index()
display(pd.DataFrame({'transactions': class_counts, 'share': class_counts / len(data)}))

ax = sns.countplot(data=data, x='Class', hue='Class', legend=False)
ax.set(title='Transaction class counts', xlabel='Class (1 = fraud)', ylabel='Transactions')
plt.show()
```

```python
display(data.groupby('Class')['Amount'].describe())

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data=data, x='Amount', hue='Class', bins=60, element='step',
             stat='density', common_norm=False, ax=axes[0])
axes[0].set(title='Amount distribution (full range)')
sns.boxplot(data=data, x='Class', y='Amount', hue='Class', showfliers=False, legend=False, ax=axes[1])
axes[1].set(title='Amount by class (outliers hidden)', xlabel='Class (1 = fraud)')
plt.tight_layout()
plt.show()
```
